```python
# TV sales Plot
plt.scatter(
    sales["TV"],
    sales["sales"],
    facecolors="none",
    edgecolors="red",
    marker="o",
    s=25
    )
plt.title("TV Sales")
plt.xlabel("spent (k $)")
plt.ylabel("sales (k units)")
plt.tight_layout
plt.show()




# Radio sales Plot
plt.scatter(
    sales["radio"],
    sales["sales"],
    facecolors="none",
    edgecolors="red",
    marker="o",
    s=25
    )
plt.title("Radio Sales")
plt.xlabel("spent (k $)")
plt.ylabel("sales (k units)")
plt.tight_layout
plt.show()



# Newspaper sales Plot
plt.scatter(
    sales["newspaper"],
    sales["sales"],
    facecolors="none",
    edgecolors="red",
    marker="o",
    s=25
    )
plt.title("Newspaper Sales")
plt.xlabel("spent (k $)")
plt.ylabel("sales (k units)")
plt.tight_layout
plt.show()
```
