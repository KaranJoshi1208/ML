Suppose you run a small business and collect this data:

| CAdvertising Spend (X, ₹k) | Sales (Y, ₹k) |
|-----------|-----------|
| 10 | 25 |
| 20 | 45 |
| 30 | 65 |
| 40 | 80 |
| 50 | 95 |


The fitted line is :  **Y = 5 + 1.8X**

```python
spend = [10, 20, 30, 40, 50]
sales = [25, 45, 65, 80, 95]
n = 5

def y_hat(x):
    return 5+1.8*x
```

```python
def RSS(estimate, true):
    len = n
    rss = 0
    while(len > 0):
        len -= 1
        rss += (estimate[len] - true[len])**2
    return rss

def SXX(x):
    len = n
    sxx = 0
    x_bar = sum(x)/len
    while(len > 0):
        len -= 1
        sxx += (x[len] - x_bar)**2
    return sxx
```

```python
#  calculating error variance σ^2

estimate = [y_hat(x) for x in spend]
rss = RSS(estimate, sales)
sigma = (1/(n-2))*rss 
sxx = SXX(spend)
```

```python
var_b1hat = (sigma/sxx)**(0.5)
var_b1hat
```
